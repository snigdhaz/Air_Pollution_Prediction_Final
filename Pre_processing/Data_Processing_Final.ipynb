# New Section

```python
required_columns = [
    'From Date','To Date',
    'PM2.5 (ug/m3)','PM10 (ug/m3)',
    'NO (ug/m3)', 'NH3 (ug/m3)',
    'NO2 (ug/m3)','NOx (ppb)',
    'SO2 (ug/m3)','CO (mg/m3)',
    'Ozone (ug/m3)', 'BP (mmHg)',
    'WS (m/s)',
    'AT (degree C)','RH (%)', 'RF (mm)'
]
```

```python
column_mapping = {
    'From Date': 'from_date',
    'To Date': 'to_date',
    'PM2.5 (ug/m3)': 'pm25',
    'PM10 (ug/m3)': 'pm10',
    'NO (ug/m3)': 'no',
    'NH3 (ug/m3)': 'nh3',
    'NO2 (ug/m3)': 'no2',
    'NOx (ppb)': 'nox',
    'SO2 (ug/m3)': 'so2',
    'CO (mg/m3)': 'co',
    'Ozone (ug/m3)': 'ozone',
    'BP (mmHg)': 'bp',
    'WS (m/s)': 'wind_speed',
    'AT (degree C)': 'air_temp',
    'RH (%)': 'humidity',
    'RF (mm)': 'rainfall'
}
```

```python

cities=['Delhi', 'Mumbai', 'Bengaluru', 'Howrah', 'Hyderabad']
city_file_names = {city: [] for city in cities }


stations_df = pd.read_csv('stations_info.csv')

for _, row in stations_df.iterrows():
  if row['city'] in cities and row['start_year'] <=2019:
    city_file_names[row['city']].append(row['file_name'])


print(city_file_names)
```

```python
missing_columns_log = []

for city, city_file in city_file_names.items():
  for x in city_file:
    x = x + ".csv"
    df = pd.read_csv(x)

    missing_cols = [
                  col for col in required_columns if col not in df.columns
              ]
    if missing_cols:
                  missing_columns_log.append({
                      'file': x,
                      'city': city,
                      'missing_columns': missing_cols
                  })
                  print(f"Skipping {x}: missing columns {missing_cols}")
    else:
      print(f"All data present for file: {x}")

```

```python
final_dfs ={city: [] for city in cities}
for city, city_file in city_file_names.items():
  df_list = []

  for x in city_file:
    x = x+ ".csv"
    df = pd.read_csv(x)

      # Find which required columns exist in this file
    existing_cols = [col for col in required_columns if col in df.columns]

    if existing_cols:
          # Select only existing required columns
        df_subset = df[existing_cols].copy()

        # Rename them using mapping
        df_subset.rename(columns=column_mapping, inplace=True)

        df_subset['city'] = city

        df_list.append(df_subset)

  # Concatenate vertically (row-wise)
  final_df = pd.concat(df_list, ignore_index=True)

  print("Final shape:", final_df.shape)
  print(final_df.head())

  final_dfs[city].append(final_df)
```

```python
for city, data in final_dfs.items():
  if city == "Delhi":
    df_city= data[0]
    df_city.to_csv('dalhi_final_dataset.csv', index=False)
```

```python
for city, data in final_dfs.items():
  df_city = data[0]
  df_city['from_date'] = pd.to_datetime(df_city['from_date'])
  df_city['to_date'] = pd.to_datetime(df_city['to_date'])

  grouped_df = (
      df_city
      .groupby(['from_date', 'to_date'], as_index=False)
      .mean(numeric_only=True)
  )

  #grouped_df.head()
  df_new = grouped_df[grouped_df['from_date'].dt.year >=2019]
  file_name = city + "_raw.csv"
  df_new.to_csv(file_name,  index=False)
  print(f"City: {city}, Shape: {df_new.shape}, NaN info: {df_new.isna().sum()}")
```
