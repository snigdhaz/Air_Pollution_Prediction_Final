```python
import numpy as np
import matplotlib.pyplot as plt

pollutants = ['PM2.5', 'NO2', 'CO', 'Ozone']

model1 = [29.5039, 9.9716, 0.3292, 11.0199 ]
model2 = [30.394, 10.570, 0.334, 10.828]
model3 = [24.7520, 8.0518, 0.2873, 8.7423]

x = np.arange(len(pollutants))
width = 0.25

plt.figure()

plt.bar(x - width, model1, width, label='Stacked Bi-LSTM with Attention (SOTA)', color='darkturquoise')
plt.bar(x, model2, width, label='Enhanced Bi-LSTM with TCN (SOTA)', color='lightgrey' )
plt.bar(x + width, model3, width, label='LSTM',color='royalblue')

plt.xticks(x, pollutants)
plt.xlabel('Pollutants')
plt.ylabel('RMSE')
plt.title('Comparison Across Models - MTL')
plt.legend()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

pollutants = ['PM2.5', 'NO2', 'CO', 'Ozone']

model1 = [29.5039, 9.9716, 0.3292, 11.0199 ]
model2 = [30.394, 10.570, 0.334, 10.828]
model3 = [24.7520, 8.0518, 0.2873, 8.7423]
model4 = [23.6255 , 8.3853 , 0.2296 , 5.9510 ]
model5 = [19.8706, 6.6539, 0.2298 , 7.1525 ]

x = np.arange(len(pollutants))
width = 0.15

plt.figure()

plt.bar(x - 2*width, model1, width, label='Stacked Bi-LSTM with Attention (SOTA)', color='#1f77b4')  # blue
plt.bar(x - width, model2, width, label='Enhanced Bi-LSTM with TCN (SOTA)', color='#ff7f0e')     # orange
plt.bar(x , model3, width, label='LSTM', color='#2ca02c')                                        # green
plt.bar(x + width, model4, width, label='Multi-Task TFT', color='#d62728')                        # red
plt.bar(x + 2*width, model5, width, label='Single-Task TFT', color='#9467bd')                        # purple

plt.xticks(x, pollutants)
plt.xlabel('Pollutants')
plt.ylabel('RMSE')
plt.title('Comparison Across Models')
plt.legend()

plt.show()
```
